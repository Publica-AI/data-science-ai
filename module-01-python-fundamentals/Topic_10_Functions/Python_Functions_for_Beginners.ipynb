# Python Functions — Beginner Notebook

## What is a function?

A function is a reusable piece of code that does a particular job.

Think of it like a small machine:

Input → Function → Output

The big idea is:

> Write the code once, then use it many times.


## 1. Creating a simple function

We use `def` to create a function.

```python
def hello():
    print("Hello!")
```

`def` means we are creating a function.

`hello` is the function's name.

The function does not run until we call it.


````python
def yes():
    print("Hello!")

yes()
````

## 2. Calling a function many times

Once we create a function, we can call it whenever we need it.


````python
def welcome():
    print("Welcome to Python!")

welcome()
welcome()
welcome()
````

````python
welcome()
````

## 3. Giving information to a function

A function can receive information.

```python
def greet(name):
```

`name` is a parameter — think of it as an empty box.

When we write `greet("John")`, `"John"` goes into that box.


````python
def greet(yes):
    print(yes)


greet("John")
greet("sahadah")


````

## 4. More than one parameter

A function can receive more than one piece of information.


````python
def add(a, b):
    print(a + b)

add(3, 5)
add(10, 20)
````

## 5. What does `return` mean?

`return` sends an answer back from the function.

```text
Give information
      ↓
   Function
      ↓
Does the job
      ↓
Returns the answer
```

This means we can store or use the answer later.


````python
def add(a, b):
    return a + b

add(5, 3)
answer = add(5,3)

print(answer*2)
````

## 6. `print()` vs `return`

`print()` displays something.

`return` sends a value back so we can use it elsewhere.

For example, a returned value can be stored in a variable or used in another calculation.


````python
def check_pass(score):
    if score >= 50:
        return "PASS"
    else:
        return "FAIL"

print(check_pass(75))
print(check_pass(40))
````

Positional Arguments

This is the simplest type of argument.

The values are matched according to their position/order.

````python
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce("Sahadah", 25)
````

````python
def subtract(a, b):
    # print(a - b)
    return a-b

subtract(10, 5)

#switch the position of the arguments
````
