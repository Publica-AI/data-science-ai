# NumPy — Beginner Notebook

## What is NumPy?

NumPy is a Python library that makes it easier to work with numbers and numerical data.

It is commonly used in Data Science, AI, and Machine Learning.

We will start with very simple examples.


## 1. Import NumPy

Before using NumPy, we import it:

```python
import numpy as np
```

`np` is simply a short name for NumPy.


````python
import numpy as np
````

````python
scores = [50, 60, 70, 80, 90]

new_scores = []

for score in scores:
    new_scores.append(score + 5)

print(new_scores)
````

````python
import numpy as np

scores = np.array([50, 60, 70, 80, 90])

new_scores = scores + 5

print(new_scores)
````

## 2. Create a NumPy array

You already know Python lists:

```python
scores = [50, 60, 70, 80, 90]
```

We can create a NumPy array using `np.array()`.

An array is useful for working with numerical data.


````python
scores = np.array([50, 60, 70, 80, 90])

print(scores)
````

````python
import numpy as np

scores = np.array([50, 60, 70, 80, 90])

print(scores + 5)
print(scores.mean())
print(scores.max())
print(scores.min())
````

## 3. Indexing

NumPy arrays use indexing like Python lists.

For this array:

```text
50 → index 0
60 → index 1
70 → index 2
80 → index 3
90 → index 4
```


````python
scores = np.array([50, 60, 70, 80, 90])

print(scores[0])
print(scores[2])
print(scores[-1])
````

## 4. Calculate on all the numbers

One useful feature is that we can perform an operation on the whole array.

If every student gets 5 extra marks:

```python
scores + 5
```

NumPy adds 5 to every score.


````python
scores = np.array([50, 60, 70, 80, 90])

new_scores = scores + 5

print(new_scores)
````

## 5. Multiplication

We can also multiply every value.

For example, if prices are doubled:


````python
prices = np.array([100, 200, 300])

new_prices = prices * 2

print(new_prices)
````

## 6. Useful NumPy calculations

Some useful operations are:

- `.sum()` → total
- `.mean()` → average
- `.max()` → highest
- `.min()` → lowest


````python
scores = np.array([50, 60, 70, 80, 90])

print("Total:", scores.sum())
print("Average:", scores.mean())
print("Highest:", scores.max())
print("Lowest:", scores.min())
````

## 7. Slicing

Slicing works like Python lists.

```python
scores[1:4]
```

means start at index 1 and stop before index 4.


````python
scores = np.array([50, 60, 70, 80, 90])

print(scores[1:4])
````

## 8. AI/ML example

Imagine these are students' study hours and exam scores.

```python
study_hours = np.array([2, 4, 6, 8, 10])
exam_scores = np.array([45, 55, 65, 75, 85])
```

NumPy lets us quickly calculate information from this numerical data.


````python
study_hours = np.array([2, 4, 6, 8, 10])
exam_scores = np.array([45, 55, 65, 75, 85])

print("Average study hours:", study_hours.mean())
print("Average exam score:", exam_scores.mean())
print("Highest exam score:", exam_scores.max())
````

## 9. Two-dimensional arrays

Think of a small spreadsheet:

```text
Study Hours    Score
4              60
6              75
8              85
```

In NumPy, we can represent this with rows and columns.


````python
students = np.array([
    [4, 60],
    [6, 75],
    [8, 85]
])

print(students)
````
